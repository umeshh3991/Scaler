Data Types

1. Numerical DataTypes

````
a = 2+5j
b = 2
c = 7.2

print(type(a))
print(type(b))
print(type(c))
````

2. Boolean Datatype

````
a = True
print(type(a))
````

3. Derived data type


i. List

````
a = [1,2,3,4.5,5]
print(type(a))
print(a)
````

Heterogeneous Example

````
a = [5,5.4,5+4j,'umesh',True]
print(type(a))
print(a)
````

````
print(a[3])
````

````
a[4] = "Sapna"
print(a)
````

ii. Tuple

````
a = (1,2,3,4.5,5)

print(type(a))
print(a)
````

Heterogeneous Example

````
a = (5,5.4,5+4j,'umesh',True)

print(type(a))
print(a)
````

Mapping DataTypes

1. Dictionaries

````
a = {"prep":"upsc",
     "optional":"math",
     "attempt":6,
     "name":"sapna"
     }

print(a)
````

````
a["name"] = "umesh"
print(a)
````
