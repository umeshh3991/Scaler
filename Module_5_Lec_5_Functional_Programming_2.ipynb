1. MAP

```python
list_1 = [1,2,3,4,5,6,7]

list_2 = [i*i for i in list_1]
list_2
```

```python
def func(x) :
  return x ** 2

list_1 = [1,2,3,4,5,6,7]
list_2 = list(map(func, list_1))
list_2
```

```python
list(map(int, "1 2 3".split(" ")))
```

```python
list(map(int, input().split(" ")))
```

```python
list_1 = [1,2,3,4,5,6,7]
print(list_1)

list(map(lambda x : x ** 2, list_1))
```

2. Filter

```python
a = list(range(1,10))
a
```

```python
b = list(filter(lambda x : x%2 == 0, a))
b
```

3. Reduce

```python
def func(x,y) :
  return x+y

def func_1(x,y) :
  return x * y

a = range(1,8)
```

```python
from functools import reduce

print(reduce(func,a))
print(reduce(func_1,a))
```

```python
reduce (lambda x,y : x+y, a)
```

```python
reduce (lambda x,y : x if x > y else y, a)
```

```python
(lambda x,y : x if x > y else y)(1,3)
```

*args and **kwargs

```python
def func_args(x,y,*args) :

  print(x)
  print(y)
  print(*args)
```

```python
a = func_args(3,4,6,7,2,1)
a
```

```python
def func_kwargs(name, age, **kwargs) :

  print(name)
  print(age)
  print(kwargs.get("gender", "Gender Not Filled"))
```

```python
func_kwargs("Umesh", 33, gender = "Male")
```

```python
func_kwargs("Sapna", 29)
```
