1.Problem Description

Write a program that returns the list of elements that are present in the given list and are divisible by 5 and 7.


Problem Constraints

1 <= A.size() <= 105
1 <= A[i] <= 105


Input Format

The only argument is an Integer Array A.


Output Format

Return an Integer Array, denoting the elements that are divisible by 5 and 7 from the given list.


Example Input

Input 1:
A = [5, 7, 70, 50, 35]
Input 2:
A = [105, 57]


Example Output

Output 1:
[70, 35]
Output 2:
[105]

```python
def divisible(n):

  A = []

  for i in (n):
    if i % 35 == 0:

      A.append(i)

  return(A)
```

```python

B = []
N = int(input("Enter number of element N : "))

for i in range (N):
  element = int(input("Enter all elemments of array : "))
  B.append(element)

Y = divisible(B)

print(Y)
```


2.Problem Description

You are given an integer array A. You have to find the second largest element/value in the array or report that no such element exists.


Problem Constraints

1 <= |A| <= 105


0 <= A[i] <= 109





Input Format

The first argument is an integer array A.





Output Format

Return the second largest element. If no such element exist then return -1.



Example Input

Input 1:

 A = [2, 1, 2]
Input 2:

 A = [2]


Example Output

Output 1:

 1
Output 2:

 -1

```python


def second_largest_element(n):

  r = 1
  while (r < len(n)):

    for i in range (len(n)-1):

      if  (n[i] < n[i+1]):

        temp = n[i+1]
        n[i+1] = n[i]
        n[i] =temp

    r += 1

  return(n)


```

```python



A = []
N = int(input("Enter number of element N : "))

for i in range(N):
  element = int(input("Enter all elemments of array : "))
  A.append(element)



Y = second_largest_element(A)

print(Y)
```


3.Problem Description

Problem Constraints

1 <= |A| <= 105
1 <= A[i] <= 109


Input Format

Single line input containing space seprated integers.
The first value in input is the length of the array A.
The following integers after the first number are the elements of the array A.


Output Format

Return an integer, which denotes the difference between maximum and minimum element of the array.


Example Input

4 1 9 10 3

3 2 3 1

```python


def difference(n):

  r = 1
  while (r < len(n)):

    for i in range (len(n)-1):

      if  (n[i] < n[i+1]):

        temp = n[i+1]
        n[i+1] = n[i]
        n[i] =temp

    r += 1

  return(n[0] - n[-1])


```

```python



A = []
N = int(input("Enter number of element N : "))

for i in range(N):
  element = int(input("Enter all elemments of array : "))
  A.append(element)



Y = difference(A)

print(Y)
```
