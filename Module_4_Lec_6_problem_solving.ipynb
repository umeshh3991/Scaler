Problem Statement:

Complete the function replace_cube(), which takes a list and a value N as inputs, and updates every element that is in a position which is a multiple of N, with a number that is the cube of the existing number, in the list and returns the newly updated list.

Note: Here while operating with N, it should be considered that if we say that an element is at index i then its position is (i+1)th. For example, an element at index 1 is actually in the second(2nd) position.

Input Format:

Space separated integers representing the list elements.
N (integer)
Output Format:

A list is simply printed
Sample Input:

1 2 3 4 5
2
Sample output:

1 8 3 64 5
Sample explanation:

Since the given value of N is 2, from the list, 2nd and 4th elements would be replaced with their cubes. The index of 2nd element is 1 but their position is (i+1) so the position is 2, which is a multiple of 2. Hence, 2 becomes 8 ad 4 becomes 64 in the resultant answer.

```python
def replace_cube(arr,N):
  output = []

  for i in range (len(arr)):
    if arr[i] % N == 0:

      arr[i] = arr[i]** 3

    output.append(arr[i])

  return arr

```

```python
list_element = []

total_element = int (input("Enter number of element : "))

factor_element = int(input("Enter factor element : "))

for i in range (total_element):

  element = int(input("Enter element : "))

  list_element.append(element)

result = replace_cube(list_element,factor_element)
print(result)
```

Given a string, complete the function modify() that returns the string after performing string modifications according to the following conditions:

1) If the length of the given string is less than 3, leave it unchanged.

2) Otherwise add 'ing' at the end of the given string.

3) If the given string already ends with 'ing' then add 'ly' instead.

Input Format:

First-line will have the number of test cases. For each testcase there will be a single line of input consisting of a string.
Output Format:

Resultant string in string format
Sample Input 1:

1
Am
Sample Output 1:

Am
Sample explanation:

The length of "Am" is lesser than 3 therefore it is returned as it is.
Sample Input 2:

1
Sow
Sample Output 2:

Sowing
Sample explanation:

The length of "Sow" is equal to 3 therefore it is returned after adding 'ing' at the end.

Sample Input 3:

1
string
Sample Output 3:

stringly
Sample explanation:

The length of "string" is greater than 3 and it has 'ing' at the end therefore it is returned after adding "ly" at the end.

```python

def modify_string(input_string):

  for i in range(len(input_string)):

    if len(input_string) <  3:
      return input_string

    elif (input_string[-3:]) != "ing":
      return input_string + "ing"


    else:
      return (input_string + "ly")



```

Given two lists that consist of different department names in two different companies, write a program that returns a set of all the departments from both companies and also all the common departments from both companies.

Input Format:

The two arguments to the function are the two lists representing the two different companies.
Output Format:

All departments and all common departments in set format.
Sample Input:

["Accounting", "Teaching", "Development", "Marketing", "Finance", "HR"]
["Marketing", "Sales", "Production", "Operations", "HR"]
Sample Output:

{'Finance', 'Teaching', 'Accounting', 'HR', 'Operations', 'Sales', 'Production', 'Marketing', 'Development'}
{'HR', 'Marketing'}
Sample explanation:

{'Finance', 'Teaching', 'Accounting', 'HR', 'Operations', 'Sales', 'Production', 'Marketing', 'Development'} are the names of all the departments present in both the companies
whereas {'HR', 'Marketing'} represents the names of departments common in both the companies.

```python
def unique_common_depart(comp1,comp2):

  unique = set(comp1 & comp2)
  common = set(comp1 | comp2)

  return(unique,common)
```

```python
total_deprt_company1 = int(input("Enter total department in company A : "))
total_deprt_company2 = int(input("Enter total department in company B : "))

dict_input1 = {
      input("department_A ")
      for i in range (total_deprt_company1)
    }
dict_input2 = {
      input("department_B ")
      for j in range (total_deprt_company2)
    }


output = unique_common_depart(dict_input1,dict_input2)

print(output)
```
