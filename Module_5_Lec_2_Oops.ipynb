Return a list with square even and odd numbers == ABSTRACTION EG

```python
class Square :

  def __init__(self, List1) :

    self.List1 = List1

  def square(self, sq_list) :

    sq_list_op = []

    for i in (sq_list) :

      i = i ** 2

      sq_list_op.append(i)

    return(sq_list_op)


  def even_number(self) :

    temp_list_1 = []

    for i in (self.List1) :

      if i % 2 == 0 :

        temp_list_1.append(i)

    return(self.square(temp_list_1))

  def odd_number(self) :

    temp_list = []

    for i in self.List1 :

      if i % 2 != 0 :

        temp_list.append(i)

    return (self.square(temp_list))
```

```python
A = [1,2,3,4,5]

C1 = Square(A)

print(C1.even_number())
print(C1.odd_number())

```

SCHOOL MGMT SYSTEM == INHERITENCE EG

```python
class Members :

  def __init__(self, name) :

    self.name = name

  def getName(self) :

    return (self.name)


class Staff(Members) :

  def __init__(self, name, salary, role) :

    super().__init__(name)
    self.salary = salary
    self.role = role

  def getSalary(self) :
    return (self.salary)

  def getRole(self) :
    return (self.role)


class Teacher(Staff) :

  def __init__(self, subject) :

    self.subject = subject

  def getSubject(self) :
    return (self.subject)



class Student(Members) :

  def __init__(self, classs, course) :

    self.classs = classs
    self.course = course

  def getClass(self) :
    return (self.classs)

  def getCourse(self) :
    return (self.course)

```

```python

stud = Staff("Sapna", "2000", "BE")

print(stud.getName())
print(stud.getSalary())
print(stud.getRole())
```
