**Functional Programming**

1. Lambda Function

```python
def square (x) :
  return(x ** 2)
```

```python
square(3)
```

```python
a = 3
(lambda x : x ** 2)(a) # anonymous function
```

```python
z = (lambda x : x ** 2)
```

```python
z(10)
```

2. If else Lambda ()

```python
la_1 = (lambda x : x if x > 5 else x+10)
```

```python
la_1(3)
```

3. Elif condition in lambda

```python
la_2 = (lambda x : x if x < 5 else (x + 10) if x < 10 else (x+20))
```

```python
la_2(6)
```

4. Multiple Arguments Using Lambda

```python
la_3 = (lambda x,y : x+y)
```

```python
la_3(2,3)
```

```python
la_4 = (lambda x,y : (x * y , x / y ))
```

```python
la_4(4,2)
```

5. Lambda Function Applications

```python
list_1 = [[11,91,3],
          [3,4,5],
          [56,89,90]]
```

```python
sorted(list_1, key = lambda x:x[1])
```

```python
dict1 = {"A":30,
         "V":20,
         "C":10,
         "D":50,
         "E":40
         }
```

```python
sorted(list(dict1.items()))
```

```python
sorted(list(dict1.items()), key = lambda x : x[1])
```

```python
output_dict = dict(sorted(list(dict1.items()), key = lambda x : x[1], reverse = True))
```

```python
print(output_dict)
```

**Higher Order Function**

```python
def hots () :
  def print_hello () :
    print("Hello World")

  return (print_hello)
```

```python
a = hots()
print(a)
```

```python
a()
```

```python
def gen_exp (n) :
  def exp (x) :
    return x ** n
  return exp
```

```python
e = gen_exp(5)

e(2)
```

```python
def add_num():
  def num(x) :
    return x
  return num
```

```python
a = add_num()
```

```python
a(5)
```
