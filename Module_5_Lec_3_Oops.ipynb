```python
class Dog :

  breed = "Shitzu"
  name = "Togo"

  def intro(self) :
    print(f"Hello my name is  {self.name} bhau bhau")
```

```python
class Cat :

  breed = "white"
  name = "Togo"
  tricks = []

  def changeName(self, new_name) :
    self.name = new_name

  def breed_trick (self, trick) :
    self.tricks.append(trick)
```

```python
c1 = Cat ()
c1.breed_trick("Fetch")
c1.breed_trick("Handshake")
```

```python
c2 = Cat()
c2.breed_trick("Jump")
```

```python
print(c1.tricks)
print(c2.tricks)
```

```python
c1.changeName("Tuffy")
c2.changeName("Toco")

print(c1.name)
print(c2.name)
```

```python
class A :
  pass

class B(A):
  pass

class C(A):
  x = 10

class D(A) :
  pass

class E(B):
  x = 5

class G(C):
  pass

class H(C):
  pass

class I(D):
  pass

class F(E) :
  pass

class J(G,H):
  pass

class K(F,J):
  pass

class L(K,I) :
  pass
```

```python
l = L()
print(l.x)
```
