Return a list with square even and odd numbers == ABSTRACTION EG

```python
class Square :

  def __init__(self, List1) :

    self.List1 = List1

  def square(self, sq_list) :

    sq_list_op = []

    for i in (sq_list) :

      i = i ** 2

      sq_list_op.append(i)

    return(sq_list_op)


  def even_number(self) :

    temp_list_1 = []

    for i in (self.List1) :

      if i % 2 == 0 :

        temp_list_1.append(i)

    return(self.square(temp_list_1))

  def odd_number(self) :

    temp_list = []

    for i in self.List1 :

      if i % 2 != 0 :

        temp_list.append(i)

    return (self.square(temp_list))
```

```python
A = [1,2,3,4,5]

C1 = Square(A)

print(C1.even_number())
print(C1.odd_number())

```

Understanding Inheritence and Polymorphism

```python
class Members :

  def __init__(self, name) :
    self.name = name

  def getName(self) :
    return(self.name)


class Teacher(Members) :

  def __init__(self, name, subject) :

    self.name = name
    self.subject = subject
```

```python
t1 = Teacher("Jatin", "OOPS")

t1.getName()
```

```python
t2 = Teacher("Jatin")
```

Use of SUPER() IN INHERITENCE

```python
class Members :

  def __init__(self, name) :
    self.name = name

  def getName(self) :
    print("Hello I am ", self.name)


class Teacher(Members) :

  def __init__(self, name, subject) :

    super().__init__(name)
    self.subject = subject


class Student(Members) :

  def __init__(self, name, course) :

    super().__init__(name)
    self.course = course
```

```python
t1 = Teacher("Jatin", "OOPS")
t1.getName()

s1 = Student("Umesh", "DSML")
s1.getName()
```
