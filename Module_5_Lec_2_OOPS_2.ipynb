Return a list with square even and odd numbers == ABSTRACTION EG

```python
class Square :

  def __init__(self, List1) :

    self.List1 = List1

  def square(self, sq_list) :

    sq_list_op = []

    for i in (sq_list) :

      i = i ** 2

      sq_list_op.append(i)

    return(sq_list_op)


  def even_number(self) :

    temp_list_1 = []

    for i in (self.List1) :

      if i % 2 == 0 :

        temp_list_1.append(i)

    return(self.square(temp_list_1))

  def odd_number(self) :

    temp_list = []

    for i in self.List1 :

      if i % 2 != 0 :

        temp_list.append(i)

    return (self.square(temp_list))
```

```python
A = [1,2,3,4,5]

C1 = Square(A)

print(C1.even_number())
print(C1.odd_number())

```

Understanding Inheritence and Polymorphism

```python
class Members :

  def __init__(self, name) :
    self.name = name

  def getName(self) :
    return(self.name)


class Teacher(Members) :

  def __init__(self, name, subject) :

    self.name = name
    self.subject = subject
```

Use of SUPER() IN INHERITENCE

```python
class Members :

  def __init__(self, name) :
    self.name = name

  def getName(self) :
    print("Hello I am ", self.name)


class Teacher(Members) :

  def __init__(self, name, subject) :

    super().__init__(name)
    self.subject = subject


class Student(Members) :

  def __init__(self, name, course) :

    super().__init__(name)
    self.course = course
```

```python
t1 = Teacher("Jatin", "OOPS")
t1.getName()

s1 = Student("Umesh", "DSML")
s1.getName()
```

Complete the classes Smaller and Larger.

Smaller class gets instantiated when the argument of the function (a:string) has a length smaller than 6 otherwise Larger is instantiated.
Both the classes have the same two methods, display() printing the type of a (class), and
evaluate() printing the number of vowels if the Name is Smaller otherwise it prints the number of consonants.

```python
class Vowel :
  def __init__(self,func_str) :
    self.func_str = func_str

  def getVowel(self) :
    count = 0
    for i in (self.func_str) :
      if i in 'AEIOUaeiou' :
        count += 1
    return(count)


class Consonant :
  def __init__(self, func_str) :
    self.func_str = func_str

  def getConsonant(self) :
    count = 0
    for i in (self.func_str) :
      if i not in 'AEIOUaeiou' :
        count += 1
    return(count)

class String_Size(Vowel, Consonant) :

  def __init__(self, func_str) :
    self.func_str = func_str

  def getString(self) :

    if (len(self.func_str)) < 6:
      print("The type of name is smaller")
      print("Vowel : ", end = "")
      return(self.getVowel())

    else :
      print("The type of name is larger")
      print("Consonant : ", end = "")
      return(self.getConsonant())
```

```python
inp_str = input("Enter the string : ")

str_obj = String_Size(inp_str)

print(str_obj.getString())
```

You are already given a class Machine_Learning with its constructor and a number of methods. You have to define a class Supervised which will inherit the MachineLearning class with the following data members and methods:


Data Members:

1. name: string
2. category: string

Methods:

1. Constructor: Initialize the data member and call the parent class constructor
2. getname(): Returns the data member name.
3. getcategory(): Returns the data member breed.
4. type(): Prints 'Supervised Learning Algorithm'.

You also have to define a class Unsupervised which will also inherit the MachineLearning class with the same data members and methods as in Supervised, but the type() will have to print Unsupervised Learning Algorithm for this class.

```python
class Machine_Learning :

  def __init__(self, name, category) :
    self.name = name
    self.category = category

  def getName(self) :
    print("Name : ", end ="")
    return(self.name)


  def getCategory(self) :
    print("Category : ", end ="")
    return(self.category)


class Supervised (Machine_Learning) :
  def __init__(self, name, category):
    super().__init__(name, category)

  def setType(self) :
    print(self.getName())
    print(self.getCategory())
    return("Supervised Learning Algorithm")



class Unsupervised (Machine_Learning) :
  def __init__(self, name, category):
    super().__init__(name, category)

  def setType(self) :
    print(self.getName())
    print(self.getCategory())
    return("Unsupervised Learning Algorithm")
```

```python
a = "Logistic Regression"
b = "Classification Algo"

c = "k-Means"
d = "Clustering Algo"

obj1=Supervised(a,b)
print(obj1.setType())

obj2=Unsupervised(c,d)
print(obj2.setType())
```
