```python
import numpy as np
import math as mt
def getEuclideandis(point1,point2):

  x1,y1 = point1
  x2,y2 = point2
  dis = ((x1-x2)**2 + (y1-y2)**2)

  return dis

def rectangularis(four_points):
  point1 = four_points[0]
  point2 = four_points[1]
  point3 = four_points[2]
  point4 = four_points[3]
  ## four side
  dis_12 = getEuclideandis(point1,point2)
  dis_23 = getEuclideandis(point2,point3)
  dis_34 = getEuclideandis(point3,point4)
  dis_41 = getEuclideandis(point4,point1)
  ###diagonal dis
  dis_13 = getEuclideandis(point1,point3)
  dis_24 = getEuclideandis(point2,point4)

  dis__arr_unique = sorted(np.unique([dis_12,dis_23,dis_34,dis_41,dis_13,dis_24]))
  print(dis__arr_unique)

  if ((dis__arr_unique[0] + dis__arr_unique[1]) == dis__arr_unique[2]):
    return True

  else:

    return False
```

Problem Description

Given 2 points A and B on the XY-Plane, complete the python function to compute the gradient of the line joining these 2 points and return the answer.

Note: The answer should be rounded of to 2 decimal places.


Input Format

Space separated values for A
Space separated values for B


Output Format

Gradient of Line Joining the 2 points (rounded of to 2 decimal places)


Example Input

1 1
2 2


Example Output

1.0

```python

def get_gradient(point):

  x1,y1 = point[0]
  x2,y2 = point[1]
  gradient = ((y1-y2) / (x1- x2))

  return gradient
```

Problem Description
You work as an expert in a construction company. Your company is asked to lay a straight road passing through 3 houses. But the company has a doubt whether these houses are in a straight line or not. So you are assigned a task to check if the 3 houses denoted by their coordinates are collinear or not.

Write a python function to check and return whether the points are collinear or not.

Input Format

Space separated values for A (x coordinates and y coordinates)
Space separated values for B (x coordinates and y coordinates)
Space separated values for C (x coordinates and y coordinates)


Output Format

True if Collinear and False if Non-Collinear


Example Input

1 2
2 2
5 4


Example Output

False

```python
def get_collinear(point):

  x1,y1 = point[0]
  x2,y2 = point[1]
  x3,y3 = point[2]
  m1 = ((y2-y1) / (x2- x1))
  m2 = ((y3-y2) / (x3- x2))

  if m1 == m2:
    return("lines are collinear ")

  else:
    return("line are not collinear ")
```

```python
p = [(1,12),(2,2),(5,4)]
print(get_collinear(p))
```
