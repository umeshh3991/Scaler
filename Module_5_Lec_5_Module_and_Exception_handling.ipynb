str(e)  provides also type of exception(ie error)

```python
li = [2, 20, 0,"reefew" , None]

for element in li:

  try:
    print("current element : ", element)
    print(5/element)

  except Exception as e:
    print("error occured : ", str(e))

  print("-"*50)
```

```python
li = [2, 20, 0,"hello" , None]

for element in li:

  try:
    print("current element : ", element)
    print(5/element)

  except ZeroDivisionError:
    print("Element is zero")

  except Exception as e:
    print("error occured : ", str(e))

  print("-"*50)
```
