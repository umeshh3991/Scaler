Create a class Bill and initialize it with the details: previous and current meter readings. Create a function total_bill() inside the Bill class which returns the total bill based on the following criteria:

1. For the first 100 units consumed: 3.5 rs/unit

2. For the next 100 units consumed: 5 rs/unit

3. For remaining units consumed: 8 rs/unit

4. Meter Charges: 150 rs (It's included only once but is mandatory)

```
class Bill_Amount () :

  def __init__(self, func_prev_read, func_curr_read, func_first_bill_flag, func_meter_charges = 150) :
    self.func_prev_read = func_prev_read
    self.func_curr_read = func_curr_read
    self.func_first_bill_flag = func_first_bill_flag
    self.func_meter_charges = func_meter_charges

  def bill_amt(self) :

    units_consumed = self.func_curr_read - self.func_prev_read

    if units_consumed <= 100 :
      func_bill_price = 3.5 * units_consumed

    elif units_consumed > 100 and units_consumed <= 200 :
      func_bill_price = 350 + 5 * (units_consumed - 100)

    else :
      func_bill_price = 850 + 8 * (units_consumed - 200)

    if (self.func_first_bill_flag == "Y") :
      func_bill_price += self.func_meter_charges

    func_bill_price = float(func_bill_price)

    return (func_bill_price)
```

```
prev_read = int(input("Please provide previous reading : "))
curr_read = int(input("Please provide current reading : "))


if (prev_read > curr_read) :
  extra_units = prev_read - curr_read
  print("You have extra", extra_units, "units left. Your bill is 0 for this month")

else :

  first_bill_flag = input("Is it a new meter and you are paying the first bill (Y/N) ?? : ")
  bill_1 = Bill_Amount(prev_read, curr_read, first_bill_flag)

  print(bill_1.bill_amt())


```
