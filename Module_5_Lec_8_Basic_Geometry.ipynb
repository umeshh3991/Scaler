Write a program to check if the graph cordinates is rectangle or not

Note : This program will not work for below 2 conditions

1. If the points provided are not in clockwise or anti-clockwise manner

```python
import math as mt

def Rectangle(points) :

  ab = mt.sqrt((points[1][0]-points[0][0])**2 + (points[1][1]-points[0][1])**2)
  cd = mt.sqrt((points[3][0]-points[2][0])**2 + (points[3][1]-points[2][1])**2)

  ad = mt.sqrt((points[3][0]-points[0][0])**2 + (points[3][1]-points[0][1])**2)
  bc = mt.sqrt((points[2][0]-points[1][0])**2 + (points[2][1]-points[1][1])**2)

  ac = mt.sqrt((points[2][0]-points[0][0])**2 + (points[2][1]-points[0][1])**2)
  bd = mt.sqrt((points[3][0]-points[1][0])**2 + (points[3][1]-points[1][1])**2)

  if ab == cd and ad == bc and ac == bd :
    return True

  return False
```

```python
points = [(2,1),(4,1),(4,3),(2,3)]

print(Rectangle(points))
```

```python
import math as mt

def EuclidienDistance(point1, point2) :

  x1, y1 = point1
  x2, y2 = point2

  return (mt.sqrt((x2-x1)**2 + (y2-y1)**2))

def isRectangle(four_points):

  pointA = four_points[0]
  pointB = four_points[1]
  pointC = four_points[2]
  pointD = four_points[3]

  distAB = EuclidienDistance(pointA, pointB)
  print(distAB)
  distCD = EuclidienDistance(pointC, pointD)
  print(distCD)

  distAD = EuclidienDistance(pointA, pointD)
  print(distAD)
  distBC = EuclidienDistance(pointB, pointC)
  print(distBC)

  diagAC = EuclidienDistance(pointA, pointC)
  print(diagAC)
  diagBD = EuclidienDistance(pointB, pointD)
  print(diagBD)

  if distAB == distCD and distAD == distBC and diagAC == diagBD :
    return True

  return False
```

```python
points = [(2,1),(7,3),(2,3),(7,1)]

print(isRectangle(points))
```
