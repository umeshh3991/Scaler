Decorator

```python
def pretty(func) :

  def inner() :

    print ("-"*50)
    func()
    print ("-"*50)

  return inner
```

```python
def foo() :

  print("Hello Everyone. How are you !")

new_foo = pretty(foo)
```

```python
new_foo()
```

```python
def bar() :

  print("I am doing good. How about you !")

new_bar = pretty(bar)
```

```python
new_bar()
```

```python
@pretty
@pretty
def new_func() :
  print("Whattttttttttttttttttt !")
```

```python
new_func()
```
