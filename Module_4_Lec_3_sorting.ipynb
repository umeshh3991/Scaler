```python
def bubble_sort(arr):
  count = 0

  for i in range(len(arr)):
    for j in range (len(arr)-1):
      count +=1

      if arr[j] > arr[j+1]:

        arr[j],arr[j+1] = arr[j+1],arr[j]

  print(count)  ##count tells about how many times iteration perform

  return(arr)

```

```python
arr =[6,7,5,8,9,1,2,3,4]

print(bubble_sort(arr))
```


Problem Description

Given an unsorted list of integers. Return the sum of the first and last elements of the list after sorting the first k elements of the list using Bubble Sort.


Input Format

Space separated values for Integers, arr
Value of k


Output Format

The required sum


Example Input

64 34 25 12 22 11 90
4


Example Output

102

```python
def bubble_sort(arr,k):
  count = 0


  for i in range(len(arr)-1):
    for j in range (k-i-1):
      count +=1


      if arr[j] > arr[j+1]:

        arr[j],arr[j+1] = arr[j+1],arr[j]

  sum = arr[0] + arr[-1]
  print(count)
  return(arr,sum)
```

```python
total_element = int(input("Enter number of element : "))

factor_element = int(input("Enter factor element : "))

list_element =[]

for i in range (total_element):

  element = int(input("Enter element : "))

  list_element.append(element)

print(bubble_sort(list_element,factor_element))
```
