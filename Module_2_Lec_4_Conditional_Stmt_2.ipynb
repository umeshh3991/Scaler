WAP ie a bartender serve drink or not he serve only when people above 25 age and bartender is not break.

```python
Age = int(input("Enter age of people : "))

Bartender_not_break = input("Whether Bartender is not on break : ")

if (Age >= 25) and Bartender_not_break != "false" and  Bartender_not_break != "False" and Bartender_not_break != "0" and Bartender_not_break != "":
  print("drink is serve")

else:
    print("drink is not serve")


```

wap two player play game rock ,paper and scissors who win print their name ,if none of win print draw the match.

```python
P1 = input("Enter P1 string : ")
P2 = input("Enter P2 string : ")

if P1 == P2:
  print("draw")

elif (P1 == "rock" and P2 == "paper") or (P1 == "paper" and P2 == "scissors") or (P1 == "paper" and P2 == "scissors"):
  print("P2 win")

else:
  print("P1 win")
```

given three no find largest no

```python
A = int(input("Enter no A : "))
B = int(input("Enter no B : "))
C = int(input("Enter no C : "))

if A >= B and A >= C:
  print("A largest no")

elif B >= C:
  print("B is largest")

else:
  print("C is largest")
```

WAP to find year is leap year or not

```python
year = int(input("Enter  year : "))

if (year % 4 == 0 ) and (year % 100 != 0)  or (year % 400 == 0):
  print("leap year")

else:
  print("not leap year")

```
