Problem Description

Write a Python function that takes a string and a length value as input arguments and returns the words from the string whose length is equal to the provided input length value.
In other words, you will split the string into individual words and then gather only those words that match the specific length given.


Input Format

The string
Value of the length of word


Output Format

A list of strings containing the words having length equal to the value provided in input.


Example Input

"The world has changed and none of us can go back all we can do is our best and sometimes the best that we can do is to start over
5


Example Output

[‘world’, ‘start’]


Example Explanation

The string has only two words with length of 5 which are "world" and "start".

```python
class check_length_equality:

  def __init__(self, string_name,length):

    self.string_name = string_name
    self.length = length


  def string1s(self):

    string_list = []

    for i in range (len(self.string_name)):
      if (len(self.string_name[i])) == self.length:

        string_list.append(self.string_name[i])

    return(string_list)

```

```python
arr = ["umesh","sapna","rahul","darsh","dfsfsd","adds"]
c1 = check_length_equality(arr,5)

print(c1.string1s())
```

```python
class student:
  def __init__(self,student_name):

    self.student_name = student_name

  def f1_stu(self):

    stu_list = sorted(self.student_name, key = lambda x : x[1])

    print(stu_list)

    return (stu_list)


class student_grade(student):

  def __init__(self , student_name):
    super(). __init__(student_name)


  def grade(self):

    obj_1 = self.f1_stu()

    for i in range (len((obj_1))):
      if obj_1[i][1] > obj_1[i-1][1]:
        break


    a = obj_1[i][1]
    b = obj_1[i][0]

    print(a)
    return(b)
```

Note: If 2 strings has the same number of 'a', keep them in the same order as entered in the list.

Input Format

Number of lists
Space separated strings for each list


Output Format

Sorted list of strings based on number of ‘a’


Example Input

3
mango banana guava
rose sandalwood
bccc aaa baac c


Example Output

[['banana', 'guava', 'mango'], ['sandalwood', 'rose'], ['aaa', 'baac', 'bccc', 'c']]

```python
class length:
  def __init__(self,string_name):

    self.string_name = string_name

  def f1_str(self):

    str_list = self.string_name

    #print(str_list,"sita")

    return (str_list)


class string_length(length):

  def __init__(self , string_name):
    super(). __init__(string_name)


  def length_a(self):
    obj_1 = self.f1_str()
    list_temp = obj_1
    #print(list_temp)

    for k in range (len(list_temp)):
      A = list_temp[0]
      B = list_temp[1]
      C = list_temp[2]


      count_list = []
      for i in range (len(A)):
        count = 0

        for j in range(len(A[i])):

          if A[i][j] == "a":
            count += 1
        count_list.append(count)

        q = (count_list)

      for k in range(len(A)):
        for j in range (k,len(A)):
          if q[k] > q[j]:

            q[k],q[j]  = q[j],q[k]
            A[k],A[j] = A[j],A[k]

      count_list = []
      for i in range (len(B)):
        count = 0

        for j in range(len(B[i])):

          if B[i][j] == "a":
            count += 1
        count_list.append(count)

        q = (count_list)

      for k in range(len(B)):
        for j in range (k,len(B)):
          if q[k] > q[j]:

            q[k],q[j]  = q[j],q[k]
            B[k],B[j] = B[j],B[k]

      count_list = []
      for i in range (len(C)):
        count = 0

        for j in range(len(C[i])):

          if C[i][j] == "a":
            count += 1
        count_list.append(count)

        q = (count_list)

      for k in range(len(C)):
        for j in range (k,len(C)):
          if q[k] > q[j]:

            q[k],q[j]  = q[j],q[k]
            C[k],C[j] = C[j],C[k]



      X = (A[::-1])
      Y = (B[::-1])
      Z = (C[::-1])

      Q = ([X] + [Y] + [Z])
      #print(Q)
      return Q
```
