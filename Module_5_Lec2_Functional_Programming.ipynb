```python
class A:
  counter = []      ##In mutable data type(like list here)wether we are using function variable(ex : self) or class variable (like : A) both are perform any function(like append,add ,sub etc)

  def __init__(self):
    self.counter.append(1)


A()
obj1 = A()

if A():
  print(A.counter)
  print(A(),A().counter)
```
