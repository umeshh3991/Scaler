1. For the first 100 units consumed: 3.5 rs/unit

2. For the next 100 units consumed: 5 rs/unit

3. For remaining units consumed: 8 rs/unit

4. Meter Charges: 150 rs (It's included only once but is mandatory)

3000.0

```python
class Bill():

  def __init__(self,prev_read,current_read,meter_charge=150):

    self.prev_read = prev_read
    self.current_read = current_read
    self.meter_charge = meter_charge

   # unit_consumed = (self.current_read - self.prev_read)


  def total_bill(self):

    unit_consumed = (self.current_read - self.prev_read)
    print(unit_consumed)

    total_charged = 0.0


    if unit_consumed < 100:

      charged1 = 3.5 * unit_consumed
      total_charged = charged1



    if  unit_consumed > 100 and unit_consumed <200:

      charged2 = 5* (unit_consumed -100)

      total_charged =  350 + charged2

    else:
      charged3 = 8* (unit_consumed - 200)

      total_charged = 850 + charged3


    #total_charged =  + charged3


    return total_charged + self.meter_charge

```

Problem Description
Create a class Student and initialize it with the following class variables:


marks in subject 1
credit of subject 1
marks in subject 2
credit of subject 2

Create a function gradepointaverage() for the Student class which computes the GPA rounded to two decimals, based on the following criteria of points in a subject :

1. If marks>=90:10 points
2. If 90>marks>=75: 9 points
3. If 75>marks>=60: 8 points
4. If 60>marks>=45: 7 points
5. If marks<45: 0 points



GPA =  
creditofsubject1+creditofsubject2
(pointsinsubject1∗creditofsubject1+pointsinsubject2∗creditofsubject2)
​


The points in a subject should be calculated according to the criteria mentioned in the question.
Also, implement a special zero case to handle when credits of both subjects are 0, return "-1.0".

Input Format

Class Student takes input in the form (marks1,marks2,credits1,credits2)


Output Format

GPA value


Example Input

(85, 76, 3, 4)


Example Output

9.0

```python
class Student():
  def __init__(self, mark1,mark2 ,credit1,credit2):


    self.mark1 = mark1
    self.mark2 = mark2
    self.credit1 = credit1
    self.credit2= credit2



  def point(self,mark):


    if mark >=90:
      return 10

    if 90 > mark >= 75:
      return 9

    if 75 > mark >= 60:
      return 8

    if 60 > mark >= 45:
      return 7

    else:
        return 0


  def gpa(self):

      if self.credit1 != 0  and self.credit2 != 0:

        grade_point_avg = ((self.point(self.mark1)) *self.credit1 + (self.point(self.mark2))*self.credit2) / (self.credit1 + self.credit2)

        return grade_point_avg



      else:

        return -1
```

```python
class square_even_number():
  def __init__(self,arr):

    self.arr = arr


  def square(self,arr):

    sq_list = []

    for i in (arr):

      i = i*i
      sq_list.append(i)

    return sq_list


  def Even_Square(self):

    temp = []

    for i in (self.arr):
      if i % 2 == 0 :

        temp.append(i)

    return (self.square(temp))


  def odd_Square(self):

    temp1 = []

    for i in (self.arr):
      if i % 2 != 0 :

        temp1.append(i)

    return (self.square(temp1))
```
