Problem Description
A colony has n people. Complete the Python function to compute how many different committees can be formed that have k number of people in it.

Input Format

Number of Test Cases
// For each case
Space separated value for Number of people in colony(N) and number of people in committee(K)


Output Format

The number of ways of forming committee for each case.


Example Input

2
10 2
5 3


Example Output

45
10

```python
import math
def committee(n,k):

  Number_of_Committee_formed = math.factorial(n)/((math.factorial(n-k))*(math.factorial(k)))

  return(Number_of_Committee_formed)
```

```python
Test_case = int(input("Enter number of test case : "))

i = 0

while i < Test_case:
  i += 1

  people =int(input("Enter number of people in colony : "))
  committee_member = int(input("Enter number of people in committee :"))

output  = committee(people, committee_member)

print(output)
```
