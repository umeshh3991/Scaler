Binary Search

```python
def binary_search(sorted_arr,target):

  start = 0
  end = len(sorted_arr)-1
  count = 0


  while start <= end:
    count += 1
    print("iteration",count)

    mid = (start +end)// 2
    if sorted_arr[mid] == target:
      return mid

    elif target > sorted_arr[mid]:
      start = mid + 1

    elif target < sorted_arr[mid]:
       end = mid - 1

  return -1

```

```python
arr = [1,3,6,8,9,76,100,420]

target = 100

print(binary_search(arr,target))
```
