```python
def binary_search(sorted_arr,target):

  start = 0
  end = len(sorted_arr)-1
  count = 0


  while start <= end:
    count += 1
    print("iteration",count)

    mid = (start +end)// 2
    if sorted_arr[mid] == target:
      return mid

    elif target > sorted_arr[mid]:
      start = mid + 1

    elif target < sorted_arr[mid]:
       end = mid - 1

  return -1

```


Given a sorted array A of size N. Find the number of elements that are less than or equal to B.

NOTE: Expected Time Complexity O(log N)

Input Format:

Space separated values for elements in array A.
An integer B.
Output Format:

Return an integer denoting the number of elements which are less than or equal to B.
Example Input:

Input 1:
A = [1, 3, 4, 4, 6]
B = 4

Input 2:
A = [1, 2, 5, 5]
B = 3
Example Output:

Output 1:
4

Output 2:
2
Example Explanation:
Explanation 1: 4 elements (1, 3, 4, 4) are less than or equal to 4.
Explanation 2: 2 elements (1, 2) are less than or equal to 3.

```python
def sorted_array(sorted_arr ,target):


  start = 0
  end = len(sorted_arr)-1
  count = 0


  while (start <= end):


    mid = (start + end) //2

    if sorted_arr[mid] <= target:
      count +=1

      start +=1

    else:

      end -=1

  return(count)


```

```python
Total_element = int(input("Enter number of element : "))

factor = int(input("Enter factor element : "))

List_element = []

for i in range (Total_element) :

  element = int(input("Enter element : "))

  List_element.append(element)

print(List_element)

output_list = sorted_array(List_element,factor)

print(output_list)

```
