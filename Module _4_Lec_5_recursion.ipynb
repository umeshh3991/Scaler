merge sort (help of recursion)

```python
def merge(arr1,arr2):
  i = 0 # iteerate over arr1
  j =0 # iterate over arr2

  result = []

  while i< len(arr1) and j< len(arr2):

    if arr1[i]< arr2[j]:
      result.append(arr1[i])
      i +=1

    else:
      result.append(arr2[j])

      j +=1

  if i < len(arr1):
      result += arr1[i:]

  if j < len(arr2):
    result += arr2[j:]

  return(result)


```

```python
def recursive_function(arr1,count1):


  for j in range(len(arr1)):

    if arr1[j] == 'h' and arr1[j+1] == 'i':
      count1 += 1

    else:

      recursive_function(arr1[j+1:],count1)


  return count1


```

Problem Description

Given an array A containing N integers.

You need to sort the array and return it.

NOTE: Try to use Insertion Sort with Recursion to solve this problem


Problem Constraints

1 <= N <= 5000

1 <= A[i] <= 103



Input Format

First and only argument is an integer array A.



Output Format

Return the sorted version of A.



Example Input

Input 1:

 A = [2, 1, 3]
Input 2:

 A = [4, 2, 1, 1]


Example Output

Output 1:

 [1, 2, 3]
Output 2:

 [1, 1, 2, 4]


Example Explanation

Explanation 1:

 The sorted version of A is [1, 2, 3]
Explanation 2:

 The sorted version of A is [1, 1, 2, 4]

```python
def insertion_rec_sort(arr,n):

  if n < 1:
    return arr


  for i in range(n):

    for j in range (i+1):

      if arr[i] < arr[j]:

       arr[i],arr[j] = arr[j], arr[i]

    #print(arr)

  return insertion_rec_sort(arr,n-1)
```

```python
arr = [4,3,3,2,1]
n = 5
print(insertion_rec_sort((arr),n))
```
