```python
def factorial(n):
  if (n == 1 or n == 0):
    return 1

  return n*factorial(n-1)
```

merge sort (help of recursion)

```python
def merge(arr1,arr2):
  i = 0 # iteerate over arr1
  j =0 # iterate over arr2

  result = []

  while i< len(arr1) and j< len(arr2):

    if arr1[i]< arr2[j]:
      result.append(arr1[i])
      i +=1

    else:
      result.append(arr2[j])

      j +=1

  if i < len(arr1):
      result += arr1[i:]

  if j < len(arr2):
    result += arr2[j:]

  return(result)


```

```python
def merge_sort(arr):

  n = len(arr)

  if n == 1:
    return(arr)

  mid = n // 2

  left = merge_sort(arr[:mid])
  right = merge_sort(arr[mid:])

  print(left,right)

  return merge(left,right)
```

sum of 1 to n munber using recursion

```python
def sum(n):

  if n == 0:
    return 0

  return n + sum(n-1)
```

write a recursion fn to find exponent given base and power

```python
def exponent (base,power):

  if base == 1 or power == 0 :
    return 1

  return (base *(exponent (base,power-1)))
```
