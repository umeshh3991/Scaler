merge sort (help of recursion)

```python
def merge(arr1,arr2):
  i = 0 # iteerate over arr1
  j =0 # iterate over arr2

  result = []

  while i< len(arr1) and j< len(arr2):

    if arr1[i]< arr2[j]:
      result.append(arr1[i])
      i +=1

    else:
      result.append(arr2[j])

      j +=1

  if i < len(arr1):
      result += arr1[i:]

  if j < len(arr2):
    result += arr2[j:]

  return(result)


```

```python
def merge_sort(arr):

  n = len(arr)

  if n == 1:
    return(arr)

  mid = n // 2

  left = merge_sort(arr[:mid])
  right = merge_sort(arr[mid:])

  print(left,right)

  return merge(left,right)
```
