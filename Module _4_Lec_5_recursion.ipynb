merge sort (help of recursion)

```python
def merge(arr1,arr2):
  i = 0 # iteerate over arr1
  j =0 # iterate over arr2

  result = []

  while i< len(arr1) and j< len(arr2):

    if arr1[i]< arr2[j]:
      result.append(arr1[i])
      i +=1

    else:
      result.append(arr2[j])

      j +=1

  if i < len(arr1):
      result += arr1[i:]

  if j < len(arr2):
    result += arr2[j:]

  return(result)


```

```python
def merge_sort(arr):

  n = len(arr)

  if n == 1:
    return(arr)

  mid = n // 2

  left = merge_sort(arr[:mid])
  right = merge_sort(arr[mid:])

  print(left,right)

  return merge(left,right)
```

Input Format:

Note: Trailing zeroes in the original number should be included in the front of reversed answer.

```python
def reverse_num(N,q):

  if N == 0:
    return q

  else:

    r = N % 10
    N = N // 10

    q = 10*q + r

    return reverse_num(N, q)






```

Problem Description

Given an array A containing N integers.

You need to sort the array and return it.

NOTE: Try to use Insertion Sort with Recursion to solve this problem


Problem Constraints

1 <= N <= 5000

1 <= A[i] <= 103



Input Format

First and only argument is an integer array A.



Output Format

Return the sorted version of A.



Example Input

Input 1:

 A = [2, 1, 3]
Input 2:

 A = [4, 2, 1, 1]


Example Output

Output 1:

 [1, 2, 3]
Output 2:

 [1, 1, 2, 4]


Example Explanation

Explanation 1:

 The sorted version of A is [1, 2, 3]
Explanation 2:

 The sorted version of A is [1, 1, 2, 4]

```python
def insertion_rec_sort(arr):

  for i in range(len(arr)):

    for j in range(i):

      if arr[i] < arr[j]:

        arr[i],arr[j] = arr[j], arr[j]

    return(arr)
```

```python
arr = [4,3,3,2,1]
print(insertion_rec_sort(arr))
```
