range ex use of for loop

```python
for x in range(3,9):

  print(x, end ="  ")
```

break ex use of for loop.

```python
for x in range(5):
   if  x == 3 :
    break

   print(x)
```

continue ex use of for loop

```python
for x in range(8):

  if x == 4:
    continue
  print(x)

else:
 print("done")
```

else use in for loop.

```python
for x in range (3,9):    ## else block only execute when iteration complete and not use when "break" in iteration but execute in "continue" statement.

  if x == 6:
    break
  print(x)
else:
  print("done")
```
