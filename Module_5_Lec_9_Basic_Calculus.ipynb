Check whether function is continuous or not ?

```python
def func(x) :
  return (x**2 + 3*x - 10)
```

```python
# LHL ==> 30

x = 5

output = []

list_x = [4.99, 4.999, 4.9999]

for x in list_x :

  y = func(x)
  output.append(y)

print(output)
```

```python
# RHL ==> 30

x = 5

output = []

list_x = [5.01, 5.001, 5.0001]

for x in list_x :

  y = func(x)
  output.append(y)

print(output)
```

Therefore, Function is continuous at x = 5

Note - If LHL is not equal to RHL then function is discontinuous

```python
def derivative(f,a) :

  h = 0.0000000000001

  d = (f(a+h) - f(a))/h

  return d

def func (x) :

  return x**2

x = 5 # 2x = 10

derivative(func, 5)
```
