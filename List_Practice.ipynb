```python
def difference(n):
  sum_even = 0
  sum_odd = 0

  for i in n:

    if i %2 == 0:
      sum_even  = sum_even + i

    else:
      sum_odd = sum_odd + i

  print(sum_even)
  print(sum_odd)

  diff =(sum_even - sum_odd)

  return(diff)

```
