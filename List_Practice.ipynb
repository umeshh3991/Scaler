```python
def duplicate(Array):

  for i in  range (len(Array)):
    if Array[i] == Array[i+1]:
      return(True)


  else:

    return(False)








```

```python
def first_multiple(list,multiplier):

  for i in (list):

    if (i // multiplier == 1):

      return(i)

  else:

    return(-1)
```

```python
Total_Elements = int(input("Enter number of element : "))

List_of_Element = []

X = int(input("Enter value of multiplier : "))

for i in range(Total_Elements):

  Element = int(input("Enter elements : "))
  List_of_Element.append(Element)

result = first_multiple(List_of_Element, X)
print(List_of_Element)

print(result)
```

```python
def second_largest_element(Array):

  #temp = 0




  for i in range (len(Array)):

    for j in range(len(Array)-i-1):

      if Array[j] <  Array[j+1]:

        temp = Array[j+1]
        Array[j+1] = Array[j]
        Array[j] =temp

  return(Array)

```
