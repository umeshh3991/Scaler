
Problem Description
Design a function check_salary() that takes salary as an argument and raises a custom exception named SalaryNotInRangeError if the salary is not in range (10000, 100000). Both the boundaries are exclusive

Handle the raised exception with the message "Salary is not in range" else print "Congratulations!!".

Input Format

A numerical value representing the salary


Output Format

The output message as per the question


Example Input

5000 #first example's input
25000 #second example's input


Example Output

"Salary is not in range" #output for first example
"Congratulations!!" #output for second example

```python
class SalaryNotInRangeError(Exception):
    """
    Exception raised for errors in the input salary.
    Attributes:
        salary -- input salary which caused the error
        message -- explanation of the error
    """
    def __init__(self, salary, message="Salary is not in range"):
        self.salary = salary
        self.message = message
       # super().__init__(self.message)


def check_salary(salary):
    try:
        if not (10000 < salary < 100000):
            raise SalaryNotInRangeError(salary) # Pass the salary here
        else:
            print("Congratulations!!")

    except SalaryNotInRangeError as e:
        print(e.message) # Print the custom message
```

```python
# Test with values at the boundaries and within/outside the range

print("Testing with salary = 10000:")
check_salary(10000)

print("\nTesting with salary = 10001:")
check_salary(10001)

print("\nTesting with salary = 50000:")
check_salary(50000)

print("\nTesting with salary = 99999:")
check_salary(99999)

print("\nTesting with salary = 100000:")
check_salary(100000)

print("\nTesting with salary = 150000:")
check_salary(150000)
```

Problem Description
Complete the functions divide(), multiply() and subtract() which get two inputs (integers a and b). These functions are called element-wise inside the calci() function for it's two inputs (list1 and list2). The tasks for the mentioned functions are explained below:
1. Divide -> integer division, ex: 4//3 = 1
2. Multiply -> multiplication, ex: 4*3 = 12
3. Subtract -> subtraction, ex: 4-3 = 1

In the above-mentioned examples, 4 and 2 can be considered as the first elements of list1 and list2 respectively. All the above-mentioned functions are expected to return integers.

Note: Make sure to handle the exception ZeroDivisionError and return it in case encountered.

Input Format

Space separated integers in two lines representing 2 Lists as argument to the function.


Output Format

(Divide, Multiply, Subtract)
# Divide should return an integer value. In case of division by zero, an exception should be raised
# Multiply function should return the multiplication of the 2 numbers
# Subtract function should return difference of the 2 numbers


Example Input

4 1 3
2 1 1


Example Output

([2,1,3], [8,1,3], [2,0,2])

```python
def divide(a,b):

    c = a / b
    return round(c ,2)

def mul(a,b):

    c = a * b
    return c

def sub(a,b):

    c = a - b
    return c

def calculate(l1,l2):
  arr1 = []
  arr2 = []
  arr3 = []


  if len(l1) != len(l2):
    return("length of both list are not same,error comes")

  else:

    for i in range (0,len((l1))):

      arr1.append(divide(l1[i],l2[i]))

      arr2.append(mul(l1[i],l2[i]))

      arr3.append(sub(l1[i],l2[i]))


    return arr1,arr2,arr3
```

```python
l1 = [4,7,3,5]
l2 = [6,3,1]
result = calculate(l1,l2)
print(result)
```


Problem Description
You have been provided with the coefficients of a quadratic equation (ax2+bx+c=0). Find the values of x for which the given quadratic equation attains a minimum value.

Input Format

Space Separated values for a,b,c (Coefficients of ax2 + bx + c = 0)
Space Separated value for lower limit and upper limit of variable x


Output Format

The Integer value of x for which the quadratic equation attains minimum value


Example Input

2 3 -1
1 3


Example Output

1

```python
def find_quadratic_minimum(a, b, c, lowerlimit, upperlimit):
    # Calculate the x-coordinate of the vertex
    # The minimum of ax^2 + bx + c occurs at x = -b / (2a)
    if a == 0:
        # If a is 0, it's a linear equation, the minimum is at the lower or upper limit
        if b >= 0:
            return lowerlimit
        else:
            return upperlimit
    else:
        vertex_x = -b / (2 * a)

        # Check if the vertex is within the given range
        if lowerlimit <= vertex_x <= upperlimit:
            # If the vertex is within the range, the minimum is at the vertex
            return int(round(vertex_x))
        else:
            # If the vertex is outside the range, the minimum within the range
            # will be at either the lowerlimit or the upperlimit.
            # Evaluate the quadratic equation at the boundaries
            value_at_lowerlimit = a * (lowerlimit**2) + b * lowerlimit + c
            value_at_upperlimit = a * (upperlimit**2) + b * upperlimit + c

            # Return the boundary that gives the minimum value
            if value_at_lowerlimit < value_at_upperlimit:
                return lowerlimit
            else:
                return upperlimit

# Example Input
a, b, c = map(int, input("Enter coefficients a, b, c separated by spaces: ").split())
lowerlimit, upperlimit = map(int, input("Enter lower and upper limits separated by spaces: ").split())

# Find the minimum within the range
min_x = find_quadratic_minimum(a, b, c, lowerlimit, upperlimit)

# Output the integer value of x
print(min_x)
```

check number is prime or not

```python
from math import factorial

def is_prime(num):
    if num <= 1:
        return ("number is not prime number")
    for i in range(2, num):

        if num % i == 0:
            return ("number is not prime number")

    print (num ,"is prime number" )
    return "factorial of prime number :" ,factorial(num)
```

```python
y = is_prime(5)
print(y)
```
