```python
import math as m
def derivative(f,a):

  h = .0000001

  d = (f(a+h) - f(a))/h

  return d

def fn (x):
  return m.e**(x**2) + m.sin(m.log(x))*(m.cos(x))

point_at_which_derivative_calculate = 5

derivative(fn,point_at_which_derivative_calculate)
```
