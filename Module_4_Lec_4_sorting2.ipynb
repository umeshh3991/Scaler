merger sort

```python
def merger(arr1,arr2):
  i = 0 # iteerate over arr1
  j =0 # iterate over arr2

  result = []

  while i< len(arr1) and j< len(arr2):

    if arr1[i]< arr2[j]:
      result.append(arr1[i])
      i +=1

    else:
      result.append(arr2[j])

      j +=1

  if i < len(arr1):
      result += arr1[i:]


  if i < len(arr2):
      result += arr2[j:]

  return(result)


```

```python
arr1 =[3,4,5,9,12,44,104,112]
arr2 = [6,8,11,45,90]

print(merger(arr1,arr2))
```

selection_sort

```python
def sel_sort(arr):
  n =len(arr)

  for i in range(n-1):
    min_idx = i

    for j in range(i+1,n):

      if arr[min_idx] > arr[j]:
        min_idx = j

    arr[i],arr[min_idx] =arr[min_idx],arr[i]

    print(arr)
```

```python
arr = [13,52,9,16,25]

result = sel_sort(arr)

print(result)
```

```python
arr = [7,4,6,7,6,2]

result = sel_sort(arr)

print(result)
```

Given a list of numbers, perform k iterations of insertion sort on it and return the element present at the center.

Note:
For a list of length n, return the n//2th indexed element after applying k iterations of insertion sort.

Input Format

The first line contains space separated values of Integers, arr
The value of k


Output Format

The element at center after k iterations of insertion sort.


Example Input

12 11 13 5 6
2


Example Output

13


Example Explanation
The list after 2 iterations of insertion sort becomes [11, 12, 13, 5, 6]. So the central element is 13.

```python
def ins_sort(arr,k):

  for i in range(0,k):

    for j in range(i+1):

      if arr[i] < arr[j]:

        arr[i],arr[j] =arr[j],arr[i]

    mid = len(arr)//2

  print(arr)
  print(arr[mid])

  #return(arr)
```


Given a list of strings, use selection sort to sort these strings in lexicographical order.

Input Format:

space-separated strings
Output Format:

list of strings
Sample Input:

['alpha', 'gamma', 'beta', 'delta']
Sample Output:

['alpha', 'beta', 'delta', 'gamma']
Sample explanation:

alpha is lexicographically smallest as it starts with 'a' and gamma is lexicographically largest as its initial character is

```python
def sel_sort(arr):
  n =len(arr)

  for i in range(n):
    min_idx = i

    for j in range(i+1,n):

      if arr[min_idx] > arr[j]:
        min_idx = j

    arr[i],arr[min_idx] =arr[min_idx],arr[i]

  return(arr)
```

```python
arr = ['alpha','gamma','beta', 'delta']

result = sel_sort(arr)

print(result)
```
