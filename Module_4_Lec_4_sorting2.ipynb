merger sort

```python
def merger(arr1,arr2):
  i = 0 # iteerate over arr1
  j =0 # iterate over arr2

  result = []

  while i< len(arr1) and j< len(arr2):

    if arr1[i]< arr2[j]:
      result.append(arr1[i])
      i +=1

    else:
      result.append(arr2[j])

      j +=1

  if i < len(arr1):
      result += arr1[i:]


  if i < len(arr2):
      result += arr2[j:]

  return(result)


```

```python
arr1 =[3,4,5,9,12,44,104,112]
arr2 = [6,8,11,45,90]

print(merger(arr1,arr2))
```

insertion sort

```python
def ins_sort(arr):

  for i in range(1,len(arr)):

    for j in range(i+1):

      if arr[i] < arr[j]:

        arr[i],arr[j] =arr[j],arr[i]

    print(arr)

  #return(arr)
```

```python
arr = [21,13,9,8,5]

result = ins_sort(arr)

print(result)
```

```python
def ins_sort(arr):

  n = len(arr)

  for i in range(1,n):
    j = i

    while j > 0 and arr[j-1] > arr[j]:

      arr[j-1],arr[j] =arr[j],arr[j-1]

      j -=1

    print(arr)

  #return(arr)
```

```python
arr = [21,13,9,8,5]

result = ins_sort(arr)

print(result)
```

selection_sort

```python
def sel_sort(arr):
  n =len(arr)

  for i in range(n-1):
    min_idx = i

    for j in range(i+1,n):

      if arr[min_idx] > arr[j]:
        min_idx = j

    arr[i],arr[min_idx] =arr[min_idx],arr[i]

    print(arr)
```

```python
arr = [13,52,9,16,25]

result = sel_sort(arr)

print(result)
```

```python
arr = [7,4,6,7,6,2]

result = sel_sort(arr)

print(result)
```
